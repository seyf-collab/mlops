1.Read Artifact

```python
import pandas as pd

labeled_table = pd.read_csv(
    "../artifacts/labeled_table.csv"
)

print("Labeled table loaded successfully!")
print("Shape:", labeled_table.shape)
```

```python
labeled_table["order_purchase_timestamp"] = pd.to_datetime(
    labeled_table["order_purchase_timestamp"]
)

print("Minimum purchase date:",
      labeled_table["order_purchase_timestamp"].min())

print("Maximum purchase date:",
      labeled_table["order_purchase_timestamp"].max())
```

```python
print("\nLabel distribution:")
print(
    labeled_table["late_delivery"]
    .value_counts(normalize=True)
    .sort_index()
)
```

5.Sort data by date

```python
labeled_table = labeled_table.sort_values(
    "order_purchase_timestamp"
).reset_index(drop=True)
```

6.Specifying Train / Validation / Test sizes

```python
n = len(labeled_table)

train_end = int(n * 0.70)
val_end = int(n * 0.85)
```

7.Create Train / Validation / Test

```python
train = labeled_table.iloc[:train_end].copy()

validation = labeled_table.iloc[train_end:val_end].copy()

test = labeled_table.iloc[val_end:].copy()
```

8.Checking section sizes

```python
print("Train shape:", train.shape)
print("Validation shape:", validation.shape)
print("Test shape:", test.shape)
```

9.Check the range of dates for each section

```python
print("Train date range:")
print(train["order_purchase_timestamp"].min())
print(train["order_purchase_timestamp"].max())

print("\nValidation date range:")
print(validation["order_purchase_timestamp"].min())
print(validation["order_purchase_timestamp"].max())

print("\nTest date range:")
print(test["order_purchase_timestamp"].min())
print(test["order_purchase_timestamp"].max())
```

10.Checking the label distribution

Here's an important point: because we chose Time-based Split, we don't necessarily expect the label ratios to be exactly the same across partitions; we simply check and document them. This differs from Random Split, where the file requested that the label ratio be maintained across partitions.

```python
print("Train label distribution:")
print(train["late_delivery"].value_counts(normalize=True) * 100)

print("\nValidation label distribution:")
print(validation["late_delivery"].value_counts(normalize=True) * 100)

print("\nTest label distribution:")
print(test["late_delivery"].value_counts(normalize=True) * 100)
```

11.Save Artifacts

```python
train.to_csv(
    "../artifacts/train.csv",
    index=False
)

validation.to_csv(
    "../artifacts/validation.csv",
    index=False
)

test.to_csv(
    "../artifacts/test.csv",
    index=False
)

print("Train, validation, and test files saved successfully!")
```
